```python
from pathlib import Path

project_dir = Path('/content/drive/MyDrive/IDX-Exchange')
data_dir = project_dir / 'data'

print("Project folder exists:", project_dir.exists())
print("Data folder exists:", data_dir.exists())
```

```python
files = list(data_dir.iterdir())

print("Number of files:", len(files))

for file in files[:10]:
    print(file.name)
```

```python
from collections import Counter

extensions = Counter(file.suffix.lower() for file in files)

print(extensions)
```

```python
listing_files = sorted(data_dir.glob("CRMLSListing*.csv"))
sold_files = sorted(data_dir.glob("CRMLSSold*.csv"))

print("Listing files:", len(listing_files))
print("Sold files:", len(sold_files))

print("\nListing files:")
for f in listing_files:
    print(f.name)

print("\nSold files:")
for f in sold_files:
    print(f.name)
```

```python
import pandas as pd

listing_file = listing_files[0]

print("File:", listing_file.name)

df_listing = pd.read_csv(listing_file)

print("Rows:", len(df_listing))
print("Columns:", len(df_listing.columns))
```

```python
sold_file = sold_files[0]

print("File:", sold_file.name)

df_sold = pd.read_csv(sold_file)

print("Rows:", len(df_sold))
print("Columns:", len(df_sold.columns))
```

```python
listing_cols = set(df_listing.columns)
sold_cols = set(df_sold.columns)

print("Columns only in Listing:")
print(sorted(list(listing_cols - sold_cols)))

print("\nColumns only in Sold:")
print(sorted(list(sold_cols - listing_cols)))

print("\nCommon columns:", len(listing_cols & sold_cols))
```

```python
filled_sold = data_dir / "CRMLSSold202401_filled.csv"
normal_sold = data_dir / "CRMLSSold202402.csv"

df_filled = pd.read_csv(filled_sold)
df_normal = pd.read_csv(normal_sold)

print("202401 filled:")
print(df_filled.shape)

print("\n202402 normal:")
print(df_normal.shape)
```

```python
print("Filled columns:", len(df_filled.columns))
print("Normal columns:", len(df_normal.columns))

print("\nColumns only in filled:")
print(sorted(list(set(df_filled.columns) - set(df_normal.columns))))

print("\nColumns only in normal:")
print(sorted(list(set(df_normal.columns) - set(df_filled.columns))))
```
