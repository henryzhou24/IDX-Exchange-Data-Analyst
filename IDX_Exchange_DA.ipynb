**Combine, filter, and export**

```python

import pandas as pd
from pathlib import Path

# Discover the monthly source files
listing_files = sorted(data_dir.glob("CRMLSListing*.csv"))
sold_files = sorted(data_dir.glob("CRMLSSold*.csv"))

print(f"Listing files found: {len(listing_files)}")
print(f"Sold files found: {len(sold_files)}")

# Check that we have the expected monthly coverage
assert len(listing_files) == 28, (
    f"Expected 28 Listing files, found {len(listing_files)}"
)
assert len(sold_files) == 28, (
    f"Expected 28 Sold files, found {len(sold_files)}"
)

def process_monthly_files(files, dataset_name, output_filename):
    monthly_data = []
    file_counts = []

    for file in files:
        df = pd.read_csv(file)

        if "PropertyType" not in df.columns:
            raise ValueError(
                f"'PropertyType' missing from {file.name}"
            )

        rows_before = len(df)

        # Filter to the required property category
        residential = df[
            df["PropertyType"].astype("string").str.strip()
            == "Residential"
        ].copy()

        rows_after = len(residential)

        file_counts.append({
            "dataset": dataset_name,
            "filename": file.name,
            "rows_before_filter": rows_before,
            "residential_rows": rows_after,
            "rows_removed": rows_before - rows_after
        })

        monthly_data.append(residential)

    # Combine the filtered monthly files
    combined = pd.concat(monthly_data, ignore_index=True)

    # Save the combined residential dataset
    output_path = output_dir / output_filename
    combined.to_csv(output_path, index=False)

    summary = {
        "dataset": dataset_name,
        "files_processed": len(files),
        "rows_before_filter": sum(
            x["rows_before_filter"] for x in file_counts
        ),
        "residential_rows": len(combined),
        "rows_removed": sum(
            x["rows_removed"] for x in file_counts
        ),
        "columns_in_output": len(combined.columns),
        "output_file": output_filename
    }

    print(f"\n{dataset_name}")
    print("-" * 40)
    print(f"Files processed: {summary['files_processed']:,}")
    print(f"Rows before filtering: {summary['rows_before_filter']:,}")
    print(f"Residential rows: {summary['residential_rows']:,}")
    print(f"Rows removed: {summary['rows_removed']:,}")
    print(f"Output columns: {summary['columns_in_output']:,}")
    print(f"Saved to: {output_path}")

    return combined, summary, file_counts


# Process Listing files
listing_residential, listing_summary, listing_counts = (
    process_monthly_files(
        listing_files,
        "Listing",
        "CRMLSListing_202401_202604_Residential.csv"
    )
)

# Process Sold files
sold_residential, sold_summary, sold_counts = (
    process_monthly_files(
        sold_files,
        "Sold",
        "CRMLSSold_202401_202604_Residential.csv"
    )
)

# Save processing documentation
summary_df = pd.DataFrame([listing_summary, sold_summary])
counts_df = pd.DataFrame(listing_counts + sold_counts)

summary_df.to_csv(
    output_dir / "week1_processing_summary.csv",
    index=False
)
counts_df.to_csv(
    output_dir / "week1_monthly_file_counts.csv",
    index=False
)

display(summary_df)
print("\nWeek 1 processing complete.")
```

**2.2.3 Investigate duplicate IDs, listing status, and .1 columns**

```python
# 1. Check ListingKey uniqueness
print("LISTING KEY CHECK")
print("Total rows:", len(listing_df))
print("Unique ListingKey values:", listing_df["ListingKey"].nunique())
print("Missing ListingKey values:", listing_df["ListingKey"].isna().sum())
print("Repeated ListingKey rows:",
      listing_df["ListingKey"].duplicated().sum())

# 2. Inspect listing status
print("\nLISTING STATUS DISTRIBUTION")
if "MlsStatus" in listing_df.columns:
    display(
        listing_df["MlsStatus"]
        .value_counts(dropna=False)
        .head(20)
        .to_frame("count")
    )

# 3. Compare columns that have .1 suffixes
print("\nDUPLICATE-NAMED COLUMN COMPARISON")

column_pairs = [
    ("PropertyType", "PropertyType.1"),
    ("DaysOnMarket", "DaysOnMarket.1"),
    ("LivingArea", "LivingArea.1"),
    ("Latitude", "Latitude.1"),
    ("Longitude", "Longitude.1"),
    ("ListPrice", "ListPrice.1"),
    ("CloseDate", "CloseDate.1"),
    ("ListAgentFirstName", "ListAgentFirstName.1"),
    ("ListAgentLastName", "ListAgentLastName.1"),
    ("BuyerOfficeName", "BuyerOfficeName.1"),
    ("UnparsedAddress", "UnparsedAddress.1"),
]

comparison_results = []

for col_a, col_b in column_pairs:
    if col_a not in listing_df.columns or col_b not in listing_df.columns:
        continue

    a = listing_df[col_a]
    b = listing_df[col_b]

    both_present = a.notna() & b.notna()
    equal = a.eq(b) | (a.isna() & b.isna())
    disagreements = (both_present & ~equal).sum()

    comparison_results.append({
        "column_1": col_a,
        "column_2": col_b,
        "both_present_rows": int(both_present.sum()),
        "disagreements_when_both_present": int(disagreements),
        "column_1_missing": int(a.isna().sum()),
        "column_2_missing": int(b.isna().sum())
    })

display(pd.DataFrame(comparison_results))
```

**2.2.4 Investigate the anomalies**

```python
# 1. Investigate repeated ListingKey values
duplicate_rows = listing_df[
    listing_df["ListingKey"].duplicated(keep=False)
].copy()

print("DUPLICATE KEY INVESTIGATION")
print("Rows associated with repeated keys:", len(duplicate_rows))
print("Number of repeated key groups:",
      duplicate_rows["ListingKey"].nunique())

core_fields = [
    "MlsStatus", "ListPrice", "ClosePrice", "CloseDate",
    "ListingContractDate", "LivingArea", "City"
]

available_fields = [
    c for c in core_fields if c in duplicate_rows.columns
]

duplicate_variation = (
    duplicate_rows.groupby("ListingKey")[available_fields]
    .nunique(dropna=False)
)

print("Repeated-key groups with differences in core fields:",
      (duplicate_variation > 1).any(axis=1).sum())

# Show only aggregate results, not property-level records
print("\nNumber of repeated-key groups by status combinations:")
status_combinations = (
    duplicate_rows.groupby("ListingKey")["MlsStatus"]
    .agg(lambda s: " | ".join(sorted(s.dropna().astype(str).unique())))
    .value_counts()
)
display(status_combinations.to_frame("key_group_count").head(15))


# 2. Check closing fields for records marked Closed
print("\nCLOSING FIELD CHECK")

closed = listing_df[
    listing_df["MlsStatus"].astype("string").str.strip().eq("Closed")
]

print("Closed rows:", len(closed))
print("Closed rows missing ClosePrice:", closed["ClosePrice"].isna().sum())
print("Closed rows missing CloseDate:", closed["CloseDate"].isna().sum())

not_closed = listing_df[
    ~listing_df["MlsStatus"].astype("string").str.strip().eq("Closed")
]

print("\nNon-Closed rows with ClosePrice:",
      not_closed["ClosePrice"].notna().sum())
print("Non-Closed rows with CloseDate:",
      not_closed["CloseDate"].notna().sum())


# 3. Inspect the two discrepant column pairs
print("\nCOLUMN PAIR DISCREPANCIES")

for col_a, col_b in [
    ("BuyerOfficeName", "BuyerOfficeName.1"),
    ("UnparsedAddress", "UnparsedAddress.1")
]:
    a = listing_df[col_a]
    b = listing_df[col_b]

    both_present = a.notna() & b.notna()
    different = both_present & a.ne(b)
    missingness_differs = a.isna() != b.isna()

    print(f"\n{col_a} vs {col_b}")
    print("Different where both present:", int(different.sum()))
    print("Different missingness:", int(missingness_differs.sum()))
```

```python
%%writefile /content/drive/MyDrive/IDX-Exchange/week2_3_deliverable.py

from pathlib import Path
import pandas as pd

# ============================================================
# IDX Exchange Data Analyst Internship — Weeks 2–3 Deliverable
# ============================================================
# Inputs:
#   data/   : Original CRMLS monthly CSV files
#   output/ : Week 1 residential-filtered Listing and Sold CSVs
#
# Filtering logic:
#   1. Week 1 files contain records where PropertyType is
#      "Residential" (after trimming surrounding whitespace).
#   2. For the new Sold analysis dataset, retain residential
#      records, remove exact duplicate rows, and retain only
#      records with a positive ClosePrice and a valid CloseDate.
#   3. Do NOT remove repeated ListingKeys automatically:
#      different records with the same key may contain
#      meaningful differences.
#
# Raw and processed real estate data should not be uploaded
# to the public GitHub repository.
# ============================================================

PROJECT_DIR = Path("/content/drive/MyDrive/IDX-Exchange")
DATA_DIR = PROJECT_DIR / "data"
OUTPUT_DIR = PROJECT_DIR / "output"

LISTING_PATH = (
    OUTPUT_DIR / "CRMLSListing_202401_202604_Residential.csv"
)
SOLD_PATH = (
    OUTPUT_DIR / "CRMLSSold_202401_202604_Residential.csv"
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


def load_csv(path):
    """Read a CSV while avoiding mixed-type inference warnings."""
    if not path.exists():
        raise FileNotFoundError(f"Required input file not found: {path}")
    return pd.read_csv(path, low_memory=False)


# ------------------------------------------------------------
# 1. Identify property types in the original source files
# ------------------------------------------------------------

type_series = []

source_files = sorted(DATA_DIR.glob("CRMLS*.csv"))

if not source_files:
    raise FileNotFoundError(
        f"No original CRMLS CSV files found in {DATA_DIR}"
    )

for file_path in source_files:
    try:
        types_df = pd.read_csv(
            file_path,
            usecols=["PropertyType"],
            dtype="string",
            low_memory=False
        )

        values = (
            types_df["PropertyType"]
            .str.strip()
            .replace("", pd.NA)
            .dropna()
        )

        type_series.append(values)

    except ValueError:
        print(f"WARNING: PropertyType column not found in {file_path.name}")

if type_series:
    all_types = pd.concat(type_series, ignore_index=True)

    property_type_report = (
        all_types.value_counts()
        .rename_axis("PropertyType")
        .reset_index(name="row_count")
    )

    property_type_report.to_csv(
        OUTPUT_DIR / "week2_3_unique_property_types.csv",
        index=False
    )

    print("\nUnique PropertyType values found in source data:")
    print(property_type_report.to_string(index=False))
else:
    raise ValueError("Could not find PropertyType in the source files.")


# ------------------------------------------------------------
# 2. Load the Week 1 residential-filtered datasets
# ------------------------------------------------------------

listing_df = load_csv(LISTING_PATH)
sold_df = load_csv(SOLD_PATH)

for name, df in [("Listing", listing_df), ("Sold", sold_df)]:
    if "PropertyType" not in df.columns:
        raise KeyError(f"{name} dataset has no PropertyType column.")

    unexpected = (
        df["PropertyType"].astype("string").str.strip().ne("Residential")
    )

    if unexpected.any():
        raise ValueError(
            f"{name} dataset contains {int(unexpected.sum())} "
            "rows that are not PropertyType == 'Residential'."
        )


# ------------------------------------------------------------
# 3. Create a NEW filtered Sold dataset for sale-price analysis
# ------------------------------------------------------------

# Keep the Week 1 source file unchanged.
sold_filtered = sold_df.copy()

# Remove exact duplicate rows only; do not deduplicate by ListingKey.
rows_before = len(sold_filtered)
sold_filtered = sold_filtered.drop_duplicates().copy()
exact_duplicates_removed = rows_before - len(sold_filtered)

# Parse dates and prices for reliable filtering.
sold_filtered["CloseDate"] = pd.to_datetime(
    sold_filtered["CloseDate"], errors="coerce"
)

sold_filtered["ClosePrice"] = pd.to_numeric(
    sold_filtered["ClosePrice"], errors="coerce"
)

# Exclude records without a valid sale price or closing date.
valid_sale = (
    sold_filtered["ClosePrice"].gt(0)
    & sold_filtered["CloseDate"].notna()
)

invalid_sale_rows_removed = int((~valid_sale).sum())
sold_filtered = sold_filtered.loc[valid_sale].copy()

FILTERED_SOLD_PATH = (
    OUTPUT_DIR
    / "CRMLSSold_202401_202604_Residential_AnalysisFiltered.csv"
)

sold_filtered.to_csv(FILTERED_SOLD_PATH, index=False)

print("\nSold dataset filtering:")
print(f"Rows before filtering: {rows_before:,}")
print(f"Exact duplicate rows removed: {exact_duplicates_removed:,}")
print(f"Rows removed for invalid price/date: {invalid_sale_rows_removed:,}")
print(f"Rows in new filtered dataset: {len(sold_filtered):,}")
print(f"Saved to: {FILTERED_SOLD_PATH}")


# ------------------------------------------------------------
# 4. Generate null-count summary tables
# ------------------------------------------------------------

datasets = {
    "Listing": listing_df,
    "Sold": sold_df,
    "Filtered Sold": sold_filtered
}

null_summary_rows = []

for dataset_name, df in datasets.items():
    null_counts = df.isna().sum()

    for column in df.columns:
        null_count = int(null_counts[column])

        null_summary_rows.append({
            "dataset": dataset_name,
            "column": column,
            "row_count": len(df),
            "null_count": null_count,
            "non_null_count": int(len(df) - null_count),
            "null_percentage": round(
                null_count / len(df) * 100, 4
            ) if len(df) else 0
        })

null_summary = pd.DataFrame(null_summary_rows)

null_summary_path = OUTPUT_DIR / "week2_3_null_count_summary.csv"
null_summary.to_csv(null_summary_path, index=False)


# ------------------------------------------------------------
# 5. Flag every column with MORE THAN 90% missing values
# ------------------------------------------------------------

missing_over_90 = null_summary.loc[
    null_summary["null_percentage"] > 90
].copy()

missing_report_path = (
    OUTPUT_DIR / "week2_3_missing_over_90_percent.csv"
)
missing_over_90.to_csv(missing_report_path, index=False)

print("\nColumns with more than 90% missing values:")
if missing_over_90.empty:
    print("None.")
else:
    print(
        missing_over_90[
            ["dataset", "column", "null_count", "null_percentage"]
        ].to_string(index=False)
    )


# ------------------------------------------------------------
# 6. Numeric distribution summary for the required variables
# ------------------------------------------------------------

# Sale-price analysis uses the new filtered Sold dataset.
# For each metric, null/non-numeric values are excluded from
# its statistics. Other numeric values are not silently removed.

numeric_columns = [
    "ClosePrice",
    "LivingArea",
    "DaysOnMarket"
]

distribution_rows = []

for column in numeric_columns:
    if column not in sold_filtered.columns:
        print(f"WARNING: {column} not found; skipping.")
        continue

    values = pd.to_numeric(
        sold_filtered[column], errors="coerce"
    ).dropna()

    if values.empty:
        distribution_rows.append({
            "variable": column,
            "valid_count": 0,
            "min": None,
            "max": None,
            "mean": None,
            "median": None,
            "p1": None,
            "p5": None,
            "p25": None,
            "p50": None,
            "p75": None,
            "p95": None,
            "p99": None
        })
        continue

    distribution_rows.append({
        "variable": column,
        "valid_count": int(values.count()),
        "min": values.min(),
        "max": values.max(),
        "mean": values.mean(),
        "median": values.median(),
        "p1": values.quantile(0.01),
        "p5": values.quantile(0.05),
        "p25": values.quantile(0.25),
        "p50": values.quantile(0.50),
        "p75": values.quantile(0.75),
        "p95": values.quantile(0.95),
        "p99": values.quantile(0.99)
    })

distribution_summary = pd.DataFrame(distribution_rows)

distribution_path = (
    OUTPUT_DIR / "week2_3_numeric_distribution_summary.csv"
)
distribution_summary.to_csv(distribution_path, index=False)

print("\nNumeric distribution summary:")
print(distribution_summary.to_string(index=False))

print("\nAll deliverable files saved:")
for path in [
    OUTPUT_DIR / "week2_3_unique_property_types.csv",
    null_summary_path,
    missing_report_path,
    distribution_path,
    FILTERED_SOLD_PATH
]:
    print(f"- {path}")
```

```python
%%writefile -a /content/drive/MyDrive/IDX-Exchange/week2_3_deliverable.py

# ============================================================
# 7. Numeric Distribution Review: Listing and Sold
# ============================================================

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

NUMERIC_FIELDS = [
    "ClosePrice",
    "ListPrice",
    "OriginalListPrice",
    "LivingArea",
    "LotSizeAcres",
    "BedroomsTotal",
    "BathroomsTotalInteger",
    "DaysOnMarket",
    "YearBuilt"
]

# Create a separate residential analysis copy of Listing.
# Do not remove listings merely because ClosePrice is missing:
# active and pending listings may not have a closing price.
listing_analysis = listing_df.copy()

listing_analysis = listing_analysis.loc[
    listing_analysis["PropertyType"]
    .astype("string").str.strip().eq("Residential")
].drop_duplicates().copy()

listing_analysis_path = (
    OUTPUT_DIR
    / "CRMLSListing_202401_202604_Residential_AnalysisFiltered.csv"
)
listing_analysis.to_csv(listing_analysis_path, index=False)

# Rebuild the Sold analysis copy using the documented rules.
sold_analysis = sold_df.copy()

sold_analysis = sold_analysis.loc[
    sold_analysis["PropertyType"]
    .astype("string").str.strip().eq("Residential")
].drop_duplicates().copy()

sold_analysis["CloseDate"] = pd.to_datetime(
    sold_analysis["CloseDate"], errors="coerce"
)
sold_analysis["ClosePrice"] = pd.to_numeric(
    sold_analysis["ClosePrice"], errors="coerce"
)

sold_analysis = sold_analysis.loc[
    sold_analysis["ClosePrice"].gt(0)
    & sold_analysis["CloseDate"].notna()
].copy()

sold_analysis_path = (
    OUTPUT_DIR
    / "CRMLSSold_202401_202604_Residential_AnalysisFiltered.csv"
)
sold_analysis.to_csv(sold_analysis_path, index=False)

analysis_datasets = {
    "Listing": listing_analysis,
    "Sold": sold_analysis
}

# Save a fresh null-count summary for both original residential
# datasets and both final analysis datasets.
null_rows = []

for dataset_name, frame in {
    "Listing": listing_df,
    "Sold": sold_df,
    "Filtered Listing": listing_analysis,
    "Filtered Sold": sold_analysis
}.items():
    for column in frame.columns:
        null_count = int(frame[column].isna().sum())
        null_rows.append({
            "dataset": dataset_name,
            "column": column,
            "row_count": len(frame),
            "null_count": null_count,
            "non_null_count": len(frame) - null_count,
            "null_percentage": (
                round(100 * null_count / len(frame), 4)
                if len(frame) else 0
            )
        })

null_summary = pd.DataFrame(null_rows)
null_summary.to_csv(
    OUTPUT_DIR / "week2_3_null_count_summary.csv", index=False
)

missing_over_90 = null_summary.loc[
    null_summary["null_percentage"] > 90
].copy()
missing_over_90.to_csv(
    OUTPUT_DIR / "week2_3_missing_over_90_percent.csv",
    index=False
)

# Distribution summaries, IQR outlier counts, and plots.
plot_dir = OUTPUT_DIR / "week2_3_numeric_plots"
plot_dir.mkdir(parents=True, exist_ok=True)

distribution_rows = []
outlier_rows = []

for dataset_name, frame in analysis_datasets.items():
    for column in NUMERIC_FIELDS:
        if column not in frame.columns:
            print(f"WARNING: {dataset_name} is missing {column}")
            continue

        values = pd.to_numeric(
            frame[column], errors="coerce"
        ).dropna()

        if values.empty:
            print(f"WARNING: No numeric values for {dataset_name}: {column}")
            continue

        q1 = values.quantile(0.25)
        median = values.quantile(0.50)
        q3 = values.quantile(0.75)
        iqr = q3 - q1
        lower_fence = q1 - 1.5 * iqr
        upper_fence = q3 + 1.5 * iqr

        distribution_rows.append({
            "dataset": dataset_name,
            "variable": column,
            "valid_count": int(values.count()),
            "null_count": int(frame[column].isna().sum()),
            "min": values.min(),
            "max": values.max(),
            "mean": values.mean(),
            "median": median,
            "p1": values.quantile(0.01),
            "p5": values.quantile(0.05),
            "p25": q1,
            "p50": median,
            "p75": q3,
            "p95": values.quantile(0.95),
            "p99": values.quantile(0.99)
        })

        below = values < lower_fence
        above = values > upper_fence

        outlier_rows.append({
            "dataset": dataset_name,
            "variable": column,
            "q1": q1,
            "q3": q3,
            "iqr": iqr,
            "lower_iqr_fence": lower_fence,
            "upper_iqr_fence": upper_fence,
            "outliers_below": int(below.sum()),
            "outliers_above": int(above.sum()),
            "total_iqr_outliers": int(below.sum() + above.sum()),
            "smallest_value": values.min(),
            "largest_value": values.max()
        })

        # For readability, plots display the central 98% of values.
        # Full-data min/max and IQR outlier counts are retained above.
        p1 = values.quantile(0.01)
        p99 = values.quantile(0.99)
        plot_values = values.loc[values.between(p1, p99)]

        safe_name = (
            f"{dataset_name.lower().replace(' ', '_')}_{column}"
        )

        # Histogram
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.hist(plot_values, bins=40)
        ax.set_title(
            f"{dataset_name}: {column} Histogram "
            "(central 98% shown)"
        )
        ax.set_xlabel(column)
        ax.set_ylabel("Frequency")
        fig.tight_layout()
        fig.savefig(plot_dir / f"{safe_name}_histogram.png", dpi=120)
        plt.close(fig)

        # Boxplot
        fig, ax = plt.subplots(figsize=(8, 5))
        ax.boxplot(plot_values, vert=True, showfliers=False)
        ax.set_title(
            f"{dataset_name}: {column} Boxplot "
            "(central 98% shown)"
        )
        ax.set_ylabel(column)
        fig.tight_layout()
        fig.savefig(plot_dir / f"{safe_name}_boxplot.png", dpi=120)
        plt.close(fig)

distribution_summary = pd.DataFrame(distribution_rows)
distribution_summary.to_csv(
    OUTPUT_DIR / "week2_3_numeric_distribution_summary.csv",
    index=False
)

outlier_report = pd.DataFrame(outlier_rows)
outlier_report.to_csv(
    OUTPUT_DIR / "week2_3_numeric_outlier_flags.csv",
    index=False
)

print("\nAnalysis dataset row counts:")
print(f"Filtered Listing: {len(listing_analysis):,}")
print(f"Filtered Sold: {len(sold_analysis):,}")

print("\nNumeric distribution summary:")
print(distribution_summary.to_string(index=False))

print("\nIQR outlier review:")
print(outlier_report.to_string(index=False))

print("\nUpdated deliverables:")
print(f"- {listing_analysis_path}")
print(f"- {sold_analysis_path}")
print(f"- {OUTPUT_DIR / 'week2_3_null_count_summary.csv'}")
print(f"- {OUTPUT_DIR / 'week2_3_missing_over_90_percent.csv'}")
print(f"- {OUTPUT_DIR / 'week2_3_numeric_distribution_summary.csv'}")
print(f"- {OUTPUT_DIR / 'week2_3_numeric_outlier_flags.csv'}")
print(f"- {plot_dir} (histograms and boxplots)")
```

```python
import pandas as pd
from pathlib import Path

output_dir = Path("/content/drive/MyDrive/IDX-Exchange/output")

listing_path = output_dir / (
    "CRMLSListing_202401_202604_Residential_AnalysisFiltered.csv"
)
sold_path = output_dir / (
    "CRMLSSold_202401_202604_Residential_AnalysisFiltered.csv"
)
types_path = output_dir / "week2_3_unique_property_types.csv"

listing = pd.read_csv(listing_path, low_memory=False)
sold = pd.read_csv(sold_path, low_memory=False)
types = pd.read_csv(types_path)

# Standardize numeric fields used in calculations.
for col in ["ClosePrice", "ListPrice"]:
    sold[col] = pd.to_numeric(sold[col], errors="coerce")

sold["CloseDate"] = pd.to_datetime(
    sold["CloseDate"], errors="coerce"
)

print("=" * 65)
print("1. RESIDENTIAL VS. OTHER PROPERTY TYPES")
print("=" * 65)

total_records = types["row_count"].sum()
types["share_percent"] = (
    types["row_count"] / total_records * 100
).round(2)

print(types.to_string(index=False))
residential_count = types.loc[
    types["PropertyType"].eq("Residential"), "row_count"
].sum()

print(f"\nTotal source records: {total_records:,}")
print(
    f"Residential: {residential_count:,} "
    f"({residential_count / total_records * 100:.2f}%)"
)
print(
    f"Other property types: {total_records - residential_count:,} "
    f"({(total_records - residential_count) / total_records * 100:.2f}%)"
)

print("\n" + "=" * 65)
print("2. CLOSE PRICE")
print("=" * 65)

prices = sold["ClosePrice"].dropna()
print(f"Valid closing prices: {len(prices):,}")
print(f"Median: ${prices.median():,.2f}")
print(f"Mean:   ${prices.mean():,.2f}")

print("\n" + "=" * 65)
print("3. DAYS ON MARKET")
print("=" * 65)

dom = pd.to_numeric(sold["DaysOnMarket"], errors="coerce").dropna()
print(f"Valid values: {len(dom):,}")
print(f"Mean: {dom.mean():,.2f} days")
print(f"Median: {dom.median():,.2f} days")
print(f"25th percentile: {dom.quantile(0.25):,.2f}")
print(f"75th percentile: {dom.quantile(0.75):,.2f}")
print(f"95th percentile: {dom.quantile(0.95):,.2f}")
print(f"Negative values: {(dom < 0).sum():,}")
print(f"Values over 3650 days: {(dom > 3650).sum():,}")

print("\n" + "=" * 65)
print("4. SOLD ABOVE VS. BELOW LIST PRICE")
print("=" * 65)

comparable = sold.loc[
    sold["ClosePrice"].gt(0) & sold["ListPrice"].gt(0),
    ["ClosePrice", "ListPrice"]
].dropna()

above = comparable["ClosePrice"] > comparable["ListPrice"]
below = comparable["ClosePrice"] < comparable["ListPrice"]
equal = comparable["ClosePrice"] == comparable["ListPrice"]
n = len(comparable)

print(f"Sales with valid closing and list prices: {n:,}")
print(f"Above list: {above.sum():,} ({above.mean() * 100:.2f}%)")
print(f"Below list: {below.sum():,} ({below.mean() * 100:.2f}%)")
print(f"Exactly list price: {equal.sum():,} ({equal.mean() * 100:.2f}%)")

print("\n" + "=" * 65)
print("5. POSSIBLE DATE CONSISTENCY ISSUES")
print("=" * 65)

date_columns = [
    col for col in sold.columns
    if "date" in col.lower() or "timestamp" in col.lower()
]
print("Date-related columns found:")
print(date_columns)

# Prefer a recognized listing/market date if present.
date_candidates = [
    "OnMarketDate",
    "ListingContractDate",
    "ListDate",
    "ListingDate",
    "OriginalEntryTimestamp",
    "ActivationDate",
    "PurchaseContractDate"
]

found_date = False

for col in date_candidates:
    if col not in sold.columns:
        continue

    listing_dates = pd.to_datetime(sold[col], errors="coerce")
    valid = listing_dates.notna() & sold["CloseDate"].notna()

    if valid.sum() == 0:
        continue

    before = valid & (sold["CloseDate"] < listing_dates)
    print(f"\nCandidate listing date: {col}")
    print(f"Comparable rows: {valid.sum():,}")
    print(
        "CloseDate earlier than this date: "
        f"{before.sum():,} ({before.sum() / valid.sum() * 100:.2f}%)"
    )

    if before.any():
        display(sold.loc[
            before,
            [c for c in [
                "ListingKey", "City", col, "CloseDate",
                "ClosePrice", "ListPrice"
            ] if c in sold.columns]
        ].head(10))

    found_date = True
    break

if not found_date:
    print(
        "No recognized listing-date field was found automatically. "
        "Use the printed date-related columns to select the correct "
        "listing/market date before drawing a conclusion."
    )

print("\n" + "=" * 65)
print("6. COUNTIES WITH THE HIGHEST MEDIAN CLOSING PRICES")
print("=" * 65)

county_candidates = [
    "CountyOrParish", "County", "CountyName"
]
county_col = next(
    (col for col in county_candidates if col in sold.columns),
    None
)

if county_col is None:
    print("No recognized county column found.")
    print(
        "Check the column names for a county field and rerun "
        "this section with the correct name."
    )
else:
    county_data = sold.loc[
        sold["ClosePrice"].gt(0)
        & sold[county_col].notna()
    ].copy()

    county_data[county_col] = (
        county_data[county_col].astype("string").str.strip()
    )
    county_data = county_data.loc[
        county_data[county_col].ne("")
    ]

    county_summary = (
        county_data.groupby(county_col)["ClosePrice"]
        .agg(median_close_price="median", sale_count="count")
        .reset_index()
    )

    # Require at least 30 sales to avoid rankings dominated by
    # counties with only a few observations.
    county_summary = county_summary.loc[
        county_summary["sale_count"] >= 30
    ].sort_values("median_close_price", ascending=False)

    print(f"County field used: {county_col}")
    print("Top 10 counties by median closing price (at least 30 sales):")
    display(county_summary.head(10))

    county_summary.to_csv(
        output_dir / "week2_3_county_median_prices.csv",
        index=False
    )
    print(
        "Saved county summary to "
        f"{output_dir / 'week2_3_county_median_prices.csv'}"
    )
```
